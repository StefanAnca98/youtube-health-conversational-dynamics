```python
import pandas as pd
from pathlib import Path

# Define the input directories
path_scient = "Data/Videos/terms_cient/"
path_pseudo = "Data/Videos/terms_pseudo/"

def load_csvs(path):
    # Get all CSV files from the directory
    files = list(Path(path).glob("*.csv"))

    # Return an empty DataFrame if no CSV files are found
    if not files:
        print(f"No CSV files found in: {path}")
        return pd.DataFrame()

    # Read each CSV file and add a column with the source filename
    dfs = [
        pd.read_csv(file).assign(source_file=file.name)
        for file in files
    ]

    # Combine all CSV files into a single DataFrame
    return pd.concat(dfs, ignore_index=True)


# Create a dataset containing all scientific-term CSV files
df_scient = load_csvs(path_scient)

# Create a dataset containing all pseudoscientific-term CSV files
df_pseudo = load_csvs(path_pseudo)

# Display the dimensions of each resulting dataset
print("Scientific:", df_scient.shape)
print("Pseudo:", df_pseudo.shape)
```

```python
print("Scientific:")
print(df_scient["Content_Classification"].value_counts(dropna=False))

print("\nPseudoscientific:")
print(df_pseudo["Content_Classification"].value_counts(dropna=False))
```
